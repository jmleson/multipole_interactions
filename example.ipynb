# 📌 Multipole Interaction Derivation: Example Usage

This notebook demonstrates how to use the `multipole_interactions` library to:
- Derive Cartesian multipole tensors
- Compute interaction energies between two molecules
- Generate LaTeX code for human-readable derivations
- Compile results into a PDF

### 🔧 Prerequisites
- Install dependencies: `pip install sympy numpy`
- Run `python run.py` to generate initial LaTeX files
- Run `bash run_latex.sh` to compile PDFs

```python
from src.MultipoleInteraction import MultipoleInteraction
from src.Tensor import Tensor
from src.latex_helping_functions import save_lines_as_latex_file
```

## 🔹 Step 1: Generate Tensor Definitions (up to order 3)

```python
latex_doc = []
for order in range(4):  # up to hexadecapole (order 4)
    t = Tensor(order=order)
    latex_doc.append(t.to_tex(importable_tex=True))

# Save to file
save_lines_as_latex_file(
    filename="tensor_definitions.tex",
    importable_tex=False,
    content=latex_doc
)

print("✅ Tensor definitions saved to 'tensor_definitions.tex'")
```

## 🔹 Step 2: Compute Dipole–Dipole Interaction in Stacked Geometry

```python
print("🔄 Computing dipole–dipole interaction...")

# Define interaction: order 1 (dipole) × order 1 (dipole)
s = MultipoleInteraction(multipole_order_1=1, multipole_order_2=1)

# Simplify and show step-by-step derivation
s.simplify_in_latex_steps(importable_tex=False)

print("✅ Dipole–dipole derivation complete.")
```

## 🔹 Step 3: Compute Quadrupole–Quadrupole Interaction

```python
print("🔄 Computing quadrupole–quadrupole interaction...")

s_quad = MultipoleInteraction(multipole_order_1=2, multipole_order_2=2)
s_quad.simplify_in_latex_steps(importable_tex=False)

print("✅ Quadrupole–quadrupole derivation complete.")
```

## 🔹 Step 4: Compile All LaTeX Files to PDF

```python
print("🚀 Compiling LaTeX files to PDF...")

import subprocess
result = subprocess.run(["bash", "run_latex.sh"], capture_output=True, text=True)

if result.returncode == 0:
    print("✅ Success! PDFs generated in 'build/' directory.")
else:
    print("❌ Error during compilation:")
    print(result.stderr)
```
